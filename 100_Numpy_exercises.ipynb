#### 44. Consider a random 10x2 matrix representing cartesian coordinates, convert them to polar coordinates (★★☆)

````python
arr=np.random.randint(0,10,(10,2))
np.random.seed(42)
print(arr)
x,y=arr[:,0],arr[:,1]
r=np.sqrt(x**2+y**2)
t=np.arctan2(y,x)

# polar_cord=np.column_stack((r,t))
polar_cords=np.hstack((r.reshape(-1,1),t.reshape(-1,1)))
print(polar_cords)
````
